```python
import numpy as np
```

```python
X = np.array([
    [50],
    [60],
    [80],
])
```

```python
y = np.array([500, 620, 790])
```

```python
XtX = X.T.dot(X)
print("XtX:", XtX)
```

```python
XtX_reg = XtX + 0.01 * np.eye(1)
print("XtX_reg:", XtX_reg)
```

```python
w = np.linalg.inv(XtX_reg).dot(X.T).dot(y)
print("Bobot w:", w)
```

```python
luas_baru = np.array([[70]])
prediksi = luas_baru.dot(w)
print("Prediksi harga rumah 70m²:", prediksi, "juta")
```
